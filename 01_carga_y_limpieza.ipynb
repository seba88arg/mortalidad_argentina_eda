```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# ── CARGA ──────────────────────────────────────────────────────
df = pd.read_csv('/defunciones_deis.csv',
                 sep=';',
                 encoding='utf-8-sig')

print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
print(f"Total defunciones: {df['CUENTA'].sum():,}")

# ── RENOMBRAR COLUMNAS ─────────────────────────────────────────
df = df.rename(columns={
    'PROVRES'  : 'cod_provincia',
    'SEXO'     : 'cod_sexo',
    'CAUSA'    : 'cod_causa',
    'MAT'      : 'tipo_muerte',
    'GRUPEDAD' : 'grupo_edad',
    'CUENTA'   : 'defunciones'
})

# ── MAPEAR PROVINCIAS ──────────────────────────────────────────
provincias = {
    2: 'CABA', 6: 'Buenos Aires', 10: 'Catamarca',
    14: 'Córdoba', 18: 'Corrientes', 22: 'Chaco',
    26: 'Chubut', 30: 'Entre Ríos', 34: 'Formosa',
    38: 'Jujuy', 42: 'La Pampa', 46: 'La Rioja',
    50: 'Mendoza', 54: 'Misiones', 58: 'Neuquén',
    62: 'Río Negro', 66: 'Salta', 70: 'San Juan',
    74: 'San Luis', 78: 'Santa Cruz', 82: 'Santa Fe',
    86: 'Santiago del Estero', 90: 'Tucumán',
    94: 'Tierra del Fuego', 98: 'Otra/Extranjero', 99: 'Sin especificar'
}
df['provincia'] = df['cod_provincia'].map(provincias)

# ── MAPEAR SEXO ────────────────────────────────────────────────
sexo_map = {1: 'Masculino', 2: 'Femenino', 3: 'Indeterminado', 9: 'Sin especificar'}
df['sexo'] = df['cod_sexo'].map(sexo_map)

# ── LIMPIAR GRUPO EDAD ─────────────────────────────────────────
# Tiene prefijo numérico para ordenar (ej: "10.15 a 19") — lo limpiamos
df['grupo_edad_label'] = df['grupo_edad'].str.replace(r'^\d+\.', '', regex=True).str.strip()

# Conservar orden original usando el prefijo como sort key
df['grupo_edad_orden'] = df['grupo_edad'].str.extract(r'^(\d+)').astype(int)

# ── FILTRAR SOLO MUERTES TOTALES ───────────────────────────────
# MAT=T incluye todo. MAT=M es solo muertes maternas (subset).
# Para el análisis general usamos T
df_total = df[df['tipo_muerte'].isnull()].copy()

# ── GUARDAR ────────────────────────────────────────────────────
df_total.to_csv('data/processed/defunciones_clean.csv', index=False, encoding='utf-8-sig')
print(f"\nDataset limpio guardado: {df_total.shape[0]:,} filas")
print(f"Total defunciones: {df_total['defunciones'].sum():,}")
```
